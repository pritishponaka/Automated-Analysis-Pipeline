### Modified: Automated data ingestion with format safety net

**What changed:** The original block raised an error and stopped when a file was unsupported. This version
re-prompts the user (up to `MAX_ATTEMPTS` times), names the accepted formats, and validates the loaded data
before the pipeline continues.

**Conditional logic:**
- More or fewer than one file uploaded: ask again
- Extension not in the supported list: ask again, listing accepted formats
- File cannot be read: show a plain-language error and ask again
- File is empty or has duplicate column names: ask again
- All attempts used: stop with a clear error

**Why it fits this dataset:** Column names are stripped of whitespace so later references such as `Glucose`
or `D_BP` match exactly. The duplicate-row check confirms no patient record is counted twice.

```python
# @title **Upload your dataset**

from google.colab import files
import pandas as pd
import io
from pathlib import Path

#*************************************************************User Edits*******************************************************************
MAX_ATTEMPTS = 3              # how many times to ask before stopping
DROP_DUPLICATE_ROWS = False   # True = remove exact duplicate rows after loading
#*************************************************************User Edits*******************************************************************

# Supported formats from the original pipeline, each mapped to its reader.
READERS = {
    ".csv":     lambda f: pd.read_csv(f, sep=None, engine="python"),  # auto-detect delimiter
    ".tsv":     lambda f: pd.read_csv(f, sep=None, engine="python"),
    ".txt":     lambda f: pd.read_csv(f, sep=None, engine="python"),
    ".xlsx":    lambda f: pd.read_excel(f),                           # first worksheet
    ".xls":     lambda f: pd.read_excel(f),
    ".json":    lambda f: pd.read_json(f),
    ".parquet": lambda f: pd.read_parquet(f),
}
supported = ", ".join(sorted(READERS))


def explain_read_error(error, filename, extension):
    """Turn a technical exception into a short, plain-language message."""
    if isinstance(error, pd.errors.EmptyDataError):
        return f"'{filename}' appears to be empty."
    if isinstance(error, pd.errors.ParserError):
        return f"'{filename}' could not be parsed. Check for broken rows or an inconsistent delimiter."
    if isinstance(error, UnicodeDecodeError):
        return f"'{filename}' contains unreadable characters. It may not really be a {extension} file."
    return f"'{filename}' could not be read as {extension}. Details: {error}"


def validate_dataframe(data):
    """Return (cleaned_df, None) if usable, or (None, problem_message) if not."""
    data = data.copy()
    # Strip stray whitespace from headers (e.g., "Glucose " -> "Glucose")
    data.columns = data.columns.astype(str).str.strip()

    if data.empty:
        return None, "The file loaded but contains no data rows."
    if data.columns.duplicated().any():
        dupes = data.columns[data.columns.duplicated()].unique().tolist()
        return None, f"Duplicate column names found: {dupes}"
    return data, None


df = None
attempt = 0

while df is None and attempt < MAX_ATTEMPTS:
    attempt += 1
    print(f"Attempt {attempt} of {MAX_ATTEMPTS}.")
    print(f"Please upload ONE file in one of these formats: {supported}")
    uploaded = files.upload()

    # Safety net 1: exactly one file
    if len(uploaded) != 1:
        print("Please upload exactly one file.\n")
        continue

    filename = next(iter(uploaded))
    extension = Path(filename).suffix.lower()

    # Safety net 2: format must be one of the supported types
    if extension not in READERS:
        print(f"'{filename}' has an unsupported format ('{extension or 'no extension'}').")
        print(f"Please upload one of these formats: {supported}\n")
        continue

    # Safety net 3: the file must be readable
    try:
        raw = READERS[extension](io.BytesIO(uploaded[filename]))
    except Exception as e:
        print(explain_read_error(e, filename, extension))
        print("Please try a different file.\n")
        continue

    # Safety net 4: the loaded data must be usable
    df, problem = validate_dataframe(raw)
    if problem:
        print(f"{problem}\nPlease try a different file.\n")
        df = None

# Stop cleanly if every attempt failed
if df is None:
    raise ValueError(f"No valid file after {MAX_ATTEMPTS} attempts. "
                     "Re-run this cell to try again.")

# Duplicate rows: report, and optionally remove
n_dupes = int(df.duplicated().sum())
if n_dupes > 0 and DROP_DUPLICATE_ROWS:
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"Removed {n_dupes} duplicate row(s).")
elif n_dupes > 0:
    print(f"Warning: {n_dupes} duplicate row(s) found (kept).")
else:
    print("No duplicate rows found.")

print(f"\nSuccessfully loaded: {filename}")
print(f"Rows: {df.shape[0]} | Columns: {df.shape[1]}")
display(df.head())
```

### Modified: Identify known impossible values for later imputation

**Purpose:** Convert zeros that cannot be real measurements into missing values (`NaN`) so they are not treated
as data. This block does not fill them in. Imputation happens later, inside the ML pipeline.

**Why a predefined list is used:** A generic null check found no missing values (`df.info()`), because the
missing measurements were recorded as 0. Only domain knowledge reveals that a glucose, blood pressure,
skin thickness, insulin, or BMI of 0 is impossible. `Pregnancies` and `Outcome` are excluded because 0 is valid.

**What changed:**
- Added a warning for any variable name in the list that is not found in the dataset, instead of skipping it silently.
- Added a numeric-type check before replacing values.
- Added a report of how many zeros were replaced in each column.

**Interpretation:** 374 of 768 rows (48.7%) in `Insulin` and 227 (29.6%) in
`Skin_Thickness` were converted to missing, the largest share. `Glucose`, `D_BP`, and `BMI` had only a
few. Leaving these zeros in place would lower the means and distort scaling for logistic regression.

```python
# @title **Identify known impossible values for later imputation (modified)**

import numpy as np
import pandas as pd

#******************************************************User Edits****************************************************************
# Predictors where 0 is physiologically impossible (requires domain knowledge).
# Pregnancies and Outcome are excluded because 0 is valid for both.
variables_to_replace_zeros = ['Glucose', 'D_BP', 'Skin_Thickness', 'Insulin', 'BMI']
#********************************************************************************************************************************

# Safety check: warn about names that do not exist instead of skipping silently
not_found = [v for v in variables_to_replace_zeros if v not in df.columns]
if not_found:
    print(f"WARNING: these variables were not found and were skipped: {not_found}")
    print(f"Available columns: {list(df.columns)}\n")

replaced_counts = {}
for variable in variables_to_replace_zeros:
    if variable in df.columns:
        # Only replace in numeric columns; zeros in text columns mean something else
        if pd.api.types.is_numeric_dtype(df[variable]):
            replaced_counts[variable] = int((df[variable] == 0).sum())
            df[variable] = df[variable].replace(0, np.nan)
        else:
            print(f"Skipped '{variable}': not a numeric column.")

# Report how many values were converted
report = pd.DataFrame({
    "zeros_replaced": pd.Series(replaced_counts),
})
report["percent_of_rows"] = (report["zeros_replaced"] / len(df) * 100).round(1)

print("Impossible zeros replaced with missing values (NaN):")
display(report)
print("\nThe ML pipeline will impute these values using training-set medians.")
```

### Modified: Identify missing values in numeric predictors

**Purpose:** Automatically find numeric predictors and report how many values are missing in each, including
the values created by the previous block. No data is changed here. Imputation happens later in the ML pipeline.

**Conditional logic:**
- If the target variable is not in the dataset, stop with a clear error.
- If no predictor has missing values, report that none were found.
- Otherwise, display counts and percentages.
- If a predictor exceeds `HIGH_MISSING_THRESHOLD` (set to 40%), flag it.

**What changed:** Added the target check, a note on excluded non-numeric columns, percentages, sorting, and the
threshold flag.

**Results:**

| Predictor | Missing | % missing | Flag |
|---|---|---|---|
| Insulin | 374 | 48.7 | HIGH |
| Skin_Thickness | 227 | 29.6 | ok |
| D_BP | 35 | 4.6 | ok |
| BMI | 11 | 1.4 | ok |
| Glucose | 5 | 0.7 | ok |

**Interpretation:** Only `Insulin` exceeds the threshold. Its imputed values would come from just 394 real
measurements, so results that depend on `Insulin` should be read with caution. The counts match the replacement
report in the previous block exactly.

```python
# @title **Automatically identify other missing values in numeric predictors (modified)**

import pandas as pd

#******************************************************User Edits****************************************************************
target_variable = "Outcome"      # column to predict
HIGH_MISSING_THRESHOLD = 40      # % missing above which a predictor is flagged
#********************************************************************************************************************************

# Safety check: the target must exist, otherwise it would be treated as a predictor
if target_variable not in df.columns:
    raise KeyError(
        f"Target '{target_variable}' not found. Available columns: {list(df.columns)}"
    )

# Automatically identify numeric predictors (everything numeric except the target)
numeric_predictors = df.select_dtypes(include="number").columns.drop(target_variable)

# Report any non-numeric columns that are excluded from the analysis
non_numeric = [c for c in df.columns if c not in numeric_predictors and c != target_variable]
if non_numeric:
    print(f"Note: non-numeric columns excluded from predictors: {non_numeric}\n")

# Count missing values and convert to percentages
missing_counts = df[numeric_predictors].isna().sum()
predictors_with_missing = missing_counts[missing_counts > 0]

if predictors_with_missing.empty:
    print("No missing values detected in numeric predictors.")
    print("If you expected missing values, confirm the invalid-values cell ran first.")
else:
    report = pd.DataFrame({
        "Missing values": predictors_with_missing,
        "Percent missing": (predictors_with_missing / len(df) * 100).round(1),
    }).sort_values("Percent missing", ascending=False)

    # Conditional flag: heavily missing predictors deserve extra caution
    report["Flag"] = report["Percent missing"].apply(
        lambda p: f"HIGH (> {HIGH_MISSING_THRESHOLD}%)" if p > HIGH_MISSING_THRESHOLD else "ok"
    )

    print("Predictors requiring imputation:")
    display(report)

    high = report.index[report["Flag"] != "ok"].tolist()
    if high:
        print(f"\nCaution: {high} exceed {HIGH_MISSING_THRESHOLD}% missing.")
        print("Median imputation will fill these, but results depend heavily on the filled values.")
    print("\nThe ML pipeline step will fill these using training-set medians.")
```

```python
# @title **Define a function to create a boxplot and histogram combo**

#---------------------DO NOT CHANGE ANY CODE BELOW THIS LINE---------------------------------------------#
def histogram_boxplot(feature, figsize=(10,6), bins=None):
    """ Boxplot and histogram combined (horizontal boxplot version)
    feature: 1-d feature array
    figsize: size of fig (default (10,6))
    bins: number of bins (default None / auto)
    """
    import matplotlib.pyplot as plt
    import seaborn as sns
    import numpy as np
    from scipy.stats import norm

    sns.set(font_scale=1)
    f2, (ax_box2, ax_hist2) = plt.subplots(
        nrows=2,
        sharex=True,
        gridspec_kw={"height_ratios": (.25, .75)},
        figsize=figsize
    )

    # Horizontal boxplot
    sns.boxplot(x=feature, ax=ax_box2, showmeans=True, color='red')

    # Histogram
    if bins:
        sns.histplot(feature, kde=False, ax=ax_hist2, bins=bins)
    else:
        sns.histplot(feature, kde=False, ax=ax_hist2)

    # Add central tendency lines to histogram
    ax_hist2.axvline(np.mean(feature), color='black', linestyle='--', label='Mean')
    ax_hist2.axvline(np.median(feature), color='green', linestyle='-', label='Median')
    ax_hist2.axvline(feature.mode()[0], color='red', linestyle='dashed', linewidth=1, label='Mode')

    ax_hist2.legend()
    plt.tight_layout()
    plt.show()

print("Function successfully created")
```

## **Inferential Statistics**

**Purpose:** Use sample data to draw conclusions about the wider population (here, women of Pima heritage aged
21+). Each test starts from a null hypothesis (no difference or no association), computes a test statistic and
a p-value, and compares the p-value to a significance level (`alpha`, set to 0.05).

**How the decision logic works:** Every test uses an `if p_value < alpha` branch.
- If p < alpha: reject the null hypothesis (evidence of a difference or association).
- Otherwise: fail to reject the null hypothesis (insufficient evidence). This is not proof that there is no difference.

**Built-in safeguards (provided code):** Each test drops missing values first and stops with an error if there
are fewer than two observations, no variation, or the wrong group coding. Run these cells after the
invalid-zero cell so impossible zeros are excluded as missing rather than treated as measurements.

### One-sample t-test (BMI vs 30)
- **H0:** the population mean BMI equals 30. **H1:** it differs from 30.
- **Results:** n = 757, sample mean = 32.457, t = 9.764, p < 0.001.

- **Caution:** statistical significance does not by itself show clinical importance. The 11 patients with
  missing BMI were excluded.

```python
# @title **Example of a one-sample t-test**

from scipy.stats import ttest_1samp

variable = "BMI"       # CHANGE VARIABLE HERE
threshold = 30        # CHANGE REFERENCE VALUE HERE
alpha = 0.05          # CHANGE SIGNIFICANCE LEVEL HERE

#---------------------DO NOT CHANGE ANY CODE BELOW THIS LINE---------------------#

# Exclude missing observations.
values = df[variable].dropna()

if len(values) < 2 or values.std() == 0:
    raise ValueError(
        "The test requires at least two nonmissing values and nonzero variance."
    )

# Perform a two-sided one-sample t-test.
t_statistic, p_value = ttest_1samp(values, popmean=threshold)

formatted_p = "< 0.001" if p_value < 0.001 else f"= {p_value:.3f}"

if p_value < alpha:
    decision = "Reject the null hypothesis."
    direction = "higher" if t_statistic > 0 else "lower"
    interpretation = (
        f"There is evidence that the population mean {variable} "
        f"is {direction} than {threshold}."
    )
else:
    decision = "Fail to reject the null hypothesis."
    interpretation = (
        f"There is insufficient evidence that the population mean "
        f"{variable} differs from {threshold}."
    )

print(f"Null hypothesis: The population mean {variable} equals {threshold}.")
print(f"Alternative hypothesis: The population mean {variable} differs from {threshold}.")
print(f"\nNonmissing observations: {len(values)}")
print(f"Sample mean: {values.mean():.3f}")
print(f"t-statistic: {t_statistic:.3f}")
print(f"p {formatted_p}")
print(f"Decision: {decision}")
print(f"Interpretation: {interpretation}")
```

### Independent samples t-test (Welch's), BMI by Outcome
- **H0:** mean BMI is equal in the two groups. **H1:** it differs.
- **Results:**
  - Without diabetes: n = 491, mean = 30.860
  - With diabetes: n = 266, mean = 35.407
  - Mean difference (group 0 − group 1) = −4.547
  - Welch's t = −9.055, p < 0.001
- **Decision:** reject the null hypothesis.
- **Interpretation:** Patients with diabetes have a higher mean BMI than patients without it, by about 4.5
  units. The negative t-statistic reflects the order (group 0 minus group 1).
- **Caution:** this is an association in observational data, not evidence that higher BMI causes diabetes.
  Welch's version is appropriate because the group sizes differ (491 vs 266).

```python
# @title **Independent samples t-test (specific to this dataset)**

from scipy.stats import ttest_ind

variable = "BMI"
group_col = "Outcome"
alpha = 0.05

label_map = {0: "Patients without diabetes", 1: "Patients with diabetes"}

#---------------------DO NOT CHANGE ANY CODE BELOW THIS LINE---------------------#

# Exclude observations missing either the measurement or group.
test_data = df[[variable, group_col]].dropna()

if set(test_data[group_col].unique()) != {0, 1}:
    raise ValueError("The group variable must contain both groups, coded 0 and 1.")

# Set a consistent comparison order.
group1 = test_data.loc[test_data[group_col] == 0, variable]
group2 = test_data.loc[test_data[group_col] == 1, variable]

if len(group1) < 2 or len(group2) < 2:
    raise ValueError("Each group requires at least two nonmissing observations.")

if group1.var() == 0 and group2.var() == 0:
    raise ValueError("The test requires variation in at least one group.")

group1_name = label_map[0]
group2_name = label_map[1]

# Two-sided Welch's t-test.
t_statistic, p_value = ttest_ind(group1, group2, equal_var=False)

formatted_p = "< 0.001" if p_value < 0.001 else f"= {p_value:.3f}"

if p_value < alpha:
    decision = "Reject the null hypothesis."
    direction = "higher" if t_statistic > 0 else "lower"
    interpretation = (
        f"There is evidence that the population mean {variable} "
        f"for {group1_name.lower()} is {direction} than "
        f"for {group2_name.lower()}."
    )
else:
    decision = "Fail to reject the null hypothesis."
    interpretation = (
        f"There is insufficient evidence of a difference in population "
        f"mean {variable} between the two groups."
    )

print(f"Null hypothesis: The population mean {variable} is equal between groups.")
print(f"Alternative hypothesis: The population mean {variable} differs between groups.")

print(f"\n{group1_name}: n = {len(group1)}, mean = {group1.mean():.3f}")
print(f"{group2_name}: n = {len(group2)}, mean = {group2.mean():.3f}")
print(f"Mean difference (group 0 − group 1): {group1.mean() - group2.mean():.3f}")

print(f"\nWelch's independent samples t-test:")
print(f"t-statistic: {t_statistic:.3f}")
print(f"p {formatted_p}")
print(f"Decision: {decision}")
print(f"Interpretation: {interpretation}")
```

### One-way ANOVA, Glucose by Outcome
- **H0:** mean glucose is equal across groups. **H1:** at least one mean differs.
- **Results:**
  - Without diabetes: n = 497, mean = 110.644, SD = 24.777
  - With diabetes: n = 266, mean = 142.320, SD = 29.599
  - F(1, 761) = 246.518, p < 0.001
- **Decision:** reject the null hypothesis.
- **Interpretation:** Mean glucose is about 31.7 units higher in patients with diabetes. With two groups, ANOVA
  is equivalent to an equal-variance t-test (F = t²). The SDs are somewhat different (24.8 vs 29.6), so the
  equal-variance assumption is only approximate, but with these sample sizes and this size of difference the
  conclusion is not in doubt.

```python
# @title **One-way ANOVA (specific to this dataset)**

import statsmodels.api as sm
from statsmodels.formula.api import ols

variable = "Glucose"
group_col = "Outcome"
alpha = 0.05

label_map = {
    0: "Patients without diabetes",
    1: "Patients with diabetes"
}

#---------------------DO NOT CHANGE ANY CODE BELOW THIS LINE---------------------#

# Exclude observations missing the measurement or group.
test_data = df[[variable, group_col]].dropna().copy()

# Use standard column names for the model formula.
test_data.columns = ["measurement", "group"]
test_data["group"] = test_data["group"].astype("category")
test_data["group"] = test_data["group"].cat.remove_unused_categories()

group_summary = (
    test_data.groupby("group", observed=True)["measurement"]
    .agg(["count", "mean", "std"])
)

if len(group_summary) < 2:
    raise ValueError("ANOVA requires at least two groups.")

if (group_summary["count"] < 2).any():
    raise ValueError("Each group requires at least two nonmissing observations.")

if (group_summary["std"] == 0).all():
    raise ValueError("The test requires variation within at least one group.")

# Fit the model and calculate the ANOVA table.
anova_model = ols("measurement ~ C(group)", data=test_data).fit()
anova_table = sm.stats.anova_lm(anova_model, typ=2)

f_stat = anova_table.loc["C(group)", "F"]
p_value = anova_table.loc["C(group)", "PR(>F)"]

formatted_p = "< 0.001" if p_value < 0.001 else f"= {p_value:.3f}"

if p_value < alpha:
    decision = "Reject the null hypothesis."
    interpretation = (
        f"There is evidence that at least two groups differ "
        f"in population mean {variable}."
    )
else:
    decision = "Fail to reject the null hypothesis."
    interpretation = (
        f"There is insufficient evidence of a difference "
        f"in population mean {variable} across groups."
    )

# Apply readable group labels for display.
group_summary.index = [
    label_map.get(group, str(group))
    for group in group_summary.index
]

print(f"Null hypothesis: Population mean {variable} is equal across all groups.")
print(f"Alternative hypothesis: At least one population mean differs.")

print("\nGroup summaries:")
display(group_summary.round(3))

print("\nANOVA table:")
display(anova_table.round(3))

print(f"\nF-statistic: {f_stat:.3f}")
print(f"p {formatted_p}")
print(f"Decision: {decision}")
print(f"Interpretation: {interpretation}")
```

### Chi-square test of independence, Outcome vs Pregnancies > 10

| | ≤ 10 pregnancies | > 10 pregnancies |
|---|---|---|
| No diabetes (observed) | 486 | 14 |
| Diabetes (observed) | 248 | 20 |
| No diabetes (expected) | 477.865 | 22.135 |
| Diabetes (expected) | 256.135 | 11.865 |

- **H0:** `Outcome` and "Pregnancies > 10" are independent.
- **Results:** chi-square = 8.965, df = 1, p = 0.003. All expected counts are above 5, so no Fisher's exact
  warning is printed.
- **Decision:** reject the null hypothesis.
- **Interpretation:** 20 of 268 patients with diabetes (7.5%) had more than 10 pregnancies, compared with 14 of
  500 without diabetes (2.8%). That is more than expected under independence (11.9 expected).
- **Caution:** only 34 patients (4.4%) exceed the threshold, and dichotomizing a numeric variable discards
  information. `Pregnancies` and `Age` are correlated (r = 0.544), so age may partly explain the association.

```python
# @title **Chi-square test of independence (specific to this dataset)**

import pandas as pd
from scipy.stats import chi2_contingency

group_col = "Outcome"
variable = "Pregnancies"
threshold = 10
alpha = 0.05

#---------------------DO NOT CHANGE ANY CODE BELOW THIS LINE---------------------#

# Exclude missing values before creating categories.
test_data = df[[group_col, variable]].dropna().copy()
test_data["Above threshold"] = test_data[variable] > threshold

crosstab = pd.crosstab(
    test_data[group_col],
    test_data["Above threshold"]
)

if crosstab.shape[0] < 2 or crosstab.shape[1] < 2:
    raise ValueError("The test requires at least two observed categories per variable.")

# Run Pearson's chi-square test without continuity correction.
chi2_stat, p_value, dof, expected = chi2_contingency(
    crosstab, correction=False
)

formatted_p = "< 0.001" if p_value < 0.001 else f"= {p_value:.3f}"

print("Observed counts (False = ≤ threshold; True = > threshold):")
display(crosstab)

print("\nExpected counts under independence:")
display(pd.DataFrame(
    expected,
    index=crosstab.index,
    columns=crosstab.columns
).round(3))

print(f"\nNull hypothesis: {group_col} and {variable} > {threshold} are independent.")
print(f"Chi-square statistic: {chi2_stat:.3f}")
print(f"Degrees of freedom: {dof}")
print(f"p {formatted_p}")

if (expected < 5).any():
    print(
        "\nSmall expected counts detected. For this 2×2 table, "
        "consider Fisher's exact test before interpreting the chi-square result."
    )

if p_value < alpha:
    print("Decision: Reject the null hypothesis.")
    print(
        f"Interpretation: There is evidence of an association between "
        f"{group_col} and having {variable} > {threshold}, "
        "provided the test assumptions are met."
    )
else:
    print("Decision: Fail to reject the null hypothesis.")
    print(
        f"Interpretation: There is insufficient evidence of an association "
        f"between {group_col} and having {variable} > {threshold}, "
        "provided the test assumptions are met."
    )
```

**How to read the matrix:**

### Automatically preprocess, train, and evaluate ML models

**Conditional logic:** The loop builds a pipeline for each model. Every pipeline imputes with the training
median, and scaling is added **only if the model is logistic regression** (`isinstance` check). Each model is
then evaluated on the same test set.

**Confusion matrices:**

**Interpretation:**
- **Logistic regression outperformed the decision tree on every metric** except recall, where they tie at
  0.531 (43 of 81 diabetes cases found by each).
- The difference comes from false positives: the tree raised 33 false alarms against 21 for logistic regression,

```python
# @title **Automatically preprocess, train, and evaluate ML models**

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    recall_score,
    precision_score,
    f1_score,
    roc_auc_score,
    ConfusionMatrixDisplay
)

# Define the models to run automatically.
models = {
    "Logistic Regression": LogisticRegression(
        solver="liblinear",
        random_state=42
    ),
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    )
}

trained_pipelines = {}
results = []

for model_name, classifier in models.items():

    # Preprocessing is fitted using training data only.
    steps = [
        ("imputer", SimpleImputer(strategy="median"))
    ]

    # Apply scaling when appropriate for the model.
    if isinstance(classifier, LogisticRegression):
        steps.append(("scaler", StandardScaler()))

    steps.append(("classifier", classifier))

    pipeline = Pipeline(steps)
    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)
    positive_index = list(pipeline.classes_).index(1)
    probabilities = pipeline.predict_proba(X_test)[:, positive_index]

    trained_pipelines[model_name] = pipeline

    results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "F1": f1_score(y_test, predictions, zero_division=0),
        "ROC AUC": roc_auc_score(y_test, probabilities)
    })

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        predictions,
        labels=[0, 1],
        display_labels=["No Diabetes", "Diabetes"],
        cmap="Blues"
    )
    plt.title(f"{model_name}: Test Set")
    plt.tight_layout()
    plt.show()

results_df = pd.DataFrame(results).set_index("Model")

print("Test-set performance:")
display(results_df.round(3))
```

### Select a cutoff using training-set cross-validation

**Purpose:** The default cutoff of 0.5 may not suit an imbalanced outcome. This cell chooses a cutoff using
only the training data, so the test set stays untouched.

**How it works:** 5-fold stratified cross-validation produces out-of-fold probabilities for every training
patient. The cutoff that maximizes **Youden's J** (sensitivity + specificity − 1) is chosen. Imputation and
scaling are re-learned inside each fold.

**Result:** Selected cutoff = **0.443** (J = 0.505, at a sensitivity of 0.642 and a false positive rate of 0.137
in the training folds).

**Interpretation:** The selected cutoff is below 0.5, so the model flags a patient as diabetic at a lower
predicted risk. That trades some false positives for more detected cases, which fits a screening goal.
Choosing the cutoff on training data only avoids tuning on the test set.

```python
# @title **Select a cutoff using training-set cross-validation**

import numpy as np
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import roc_curve

#---------------------DO NOT CHANGE ANY CODE BELOW THIS LINE---------------------#

# Use the logistic_pipeline defined in the training block.
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

if y_train.value_counts().min() < 5:
    raise ValueError("Each training class needs at least 5 observations.")

# Each observation is predicted by a pipeline trained on other folds.
# Imputation and scaling are learned separately within each fold.
cv_probabilities = cross_val_predict(
    logistic_pipeline,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba"
)

positive_index = list(logistic_pipeline.classes_).index(1)

fpr, tpr, thresholds = roc_curve(
    y_train,
    cv_probabilities[:, positive_index]
)

# Exclude the infinite threshold representing no positive predictions.
finite = np.isfinite(thresholds)
j_scores = tpr[finite] - fpr[finite]
optimal_idx = np.argmax(j_scores)
optimal_threshold = thresholds[finite][optimal_idx]

print(f"Selected threshold: {optimal_threshold:.3f}")
print("Criterion: maximum cross-validation sensitivity + specificity − 1.")
print("Apply this cutoff to the fitted pipeline's test-set probabilities.")
```

### Build and tune decision tree using cross-validation F1

**User setting:** `scoring_metric = "f1"`. F1 balances precision and recall, which suits an imbalanced outcome
(65% negative / 35% positive). Alternatives are `"recall"`, `"precision"`, and `"accuracy"`.

**How it works:** `GridSearchCV` tries every combination of the settings below (4 × 3 × 3 × 2 = 72 combinations)
using 5-fold stratified cross-validation on the **training set only**. Median imputation is re-learned inside
each fold. The best combination is refitted on all training data.

| Setting | Values tried | Purpose |
|---|---|---|
| `max_depth` | 2, 4, 6, 8 | Limits how many questions deep the tree can go |
| `min_samples_leaf` | 5, 10, 15 | Requires each leaf to hold enough patients |
| `max_leaf_nodes` | 5, 10, 15 | Caps the number of final groups |
| `class_weight` | None, "balanced" | "balanced" gives the minority class (diabetes) more weight |

**Results:**
- Best mean cross-validation F1 = **0.682**
- Selected parameters: `class_weight = "balanced"`, `max_depth = 8`, `max_leaf_nodes = 10`, `min_samples_leaf = 5`

**Interpretation:**
- `class_weight = "balanced"` won. Penalizing missed diabetes cases more heavily improved F1, which suggests
  the class imbalance mattered.
- The fitted tree has depth **5** and **10 leaves**, even though `max_depth = 8` was allowed. The 10-leaf cap is
  what limits the tree, so `max_depth` was not the binding constraint here.
- The runners-up scored almost the same (CV F1 0.680 and 0.678), so several settings perform similarly. The
  selected combination is not clearly better than its neighbors.
- The tuned tree is far simpler than the default tree (10 leaves vs 98).

```python
# @title **Build and tune decision tree using cross-validation F1**

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold

# Choose the metric used for tuning.
scoring_metric = "f1"  # Alternatives: "recall", "precision", "accuracy"

#---------------------DO NOT CHANGE ANY CODE BELOW THIS LINE---------------------#

dtree_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("classifier", DecisionTreeClassifier(random_state=42))
])

# Pipeline parameters require the step name prefix.
parameters = {
    "classifier__max_depth": [2, 4, 6, 8],
    "classifier__min_samples_leaf": [5, 10, 15],
    "classifier__max_leaf_nodes": [5, 10, 15],
    "classifier__class_weight": [None, "balanced"]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

if y_train.value_counts().min() < 5:
    raise ValueError("Each training class needs at least 5 observations.")

grid_obj = GridSearchCV(
    estimator=dtree_pipeline,
    param_grid=parameters,
    scoring=scoring_metric,
    cv=cv,
    n_jobs=-1,
    refit=True
)

grid_obj.fit(X_train, y_train)

# Already refitted on the full training set; no additional fit is needed.
dtree = grid_obj.best_estimator_

print("Decision tree pipeline built and tuned.")
print(f"Best mean cross-validation {scoring_metric}: {grid_obj.best_score_:.3f}")
print("Selected parameters:")
print(grid_obj.best_params_)
```

### Function to calculate accuracy, recall, precision, and F1
Defines a function that scores a fitted model on both the training and test sets. It prints nothing until it is
called. The next cell reports:

| Metric | Training | Test |
|---|---|---|
| Accuracy | 0.750 | 0.719 |
| Recall | 0.909 | 0.802 |
| Precision | 0.592 | 0.570 |
| F1 | 0.717 | 0.667 |

**Interpretation:** Training and test scores are close (F1 0.717 vs 0.667, accuracy 0.750 vs 0.719), so the
tuned tree no longer overfits. The default tree had 100% training accuracy against 69.3% on the test set.
Recall is high in both sets because the tree favors predicting diabetes.

```python
# @title **Function to calculate accuracy, recall, precision, and F1**

from sklearn import metrics

def get_metrics_score(model, flag=True):
    """Evaluate a fitted classifier on the existing training and test sets."""

    pred_train = model.predict(X_train)
    pred_test = model.predict(X_test)

    train_acc = metrics.accuracy_score(y_train, pred_train)
    test_acc = metrics.accuracy_score(y_test, pred_test)

    train_recall = metrics.recall_score(
        y_train, pred_train, zero_division=0
    )
    test_recall = metrics.recall_score(
        y_test, pred_test, zero_division=0
    )

    train_precision = metrics.precision_score(
        y_train, pred_train, zero_division=0
    )
    test_precision = metrics.precision_score(
        y_test, pred_test, zero_division=0
    )

    train_f1 = metrics.f1_score(y_train, pred_train, zero_division=0)
    test_f1 = metrics.f1_score(y_test, pred_test, zero_division=0)

    score_list = [
        train_acc, test_acc,
        train_recall, test_recall,
        train_precision, test_precision,
        train_f1, test_f1
    ]

    if flag:
        print(f"{'Metric':<12} {'Training':>10} {'Test':>10}")
        for name, train_value, test_value in [
            ("Accuracy", train_acc, test_acc),
            ("Recall", train_recall, test_recall),
            ("Precision", train_precision, test_precision),
            ("F1", train_f1, test_f1)
        ]:
            print(f"{name:<12} {train_value:>10.3f} {test_value:>10.3f}")

    return score_list

print("Metric evaluation function successfully defined.")
```

```python
# @title **Evaluate the tuned decision tree on training and test sets**

# Prints training and test scores; the returned list is not needed here.
_ = get_metrics_score(dtree)
```
